```python
import cv2
import numpy as np
import os


import cv2
import numpy as np

def modify_image(img, output_path, random_pixel=True):
    modified = img.copy()
    h, w = modified.shape[:2]

    if random_pixel:
        i = np.random.randint(0, h)
        j = np.random.randint(0, w)
    else:
        i, j = h // 2, w // 2

    # Slight change
    modified[i, j] = (modified[i, j] + 1) % 256

    # ✅ Save image
    cv2.imwrite(output_path, modified)

modify_image(img = 'image.png', output_path='out_image.png')
```

```python

def calculate_npcr(img1, img2):
    assert img1.shape == img2.shape, "Images must match"

    # Count differing pixels
    diff = img1 != img2
    npcr = np.sum(diff) / diff.size * 100
    return npcr


def npcr_with_custom_encryption(input_path, encrypt_func, output_dir=None):
    
    # Read original image
    img = cv2.imread(input_path, cv2.IMREAD_UNCHANGED)
    if img is None:
        raise ValueError("Invalid image path")

    # Modify (1-pixel change)
    modified_img = modify_image(img)

    # 🔥 Apply YOUR encryption
    enc1 = encrypt_func(img)
    enc2 = encrypt_func(modified_img)

    # Save results (optional, useful for paper)
    if output_dir:
        os.makedirs(output_dir, exist_ok=True)
        cv2.imwrite(f"{output_dir}/original.png", img)
        cv2.imwrite(f"{output_dir}/modified.png", modified_img)
        cv2.imwrite(f"{output_dir}/enc_original.png", enc1)
        cv2.imwrite(f"{output_dir}/enc_modified.png", enc2)

    # Compute NPCR
    npcr_value = calculate_npcr(enc1, enc2)

    print(f"NPCR: {npcr_value:.4f}%")
    return npcr_value
```
